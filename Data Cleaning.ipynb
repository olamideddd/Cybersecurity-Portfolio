```python
import pandas as pd

# 1. Load the "Dirty" Data
df = pd.read_csv('globaltech_raw_sales.csv')
print(f"Original Row Count: {len(df)}")

# 2. Remove Duplicates
# We check for rows where ALL columns are identical and drop them.
df = df.drop_duplicates()
print(f"Row Count after removing duplicates: {len(df)}")

# 3. Standardise Text (Fixing 'laptop' vs 'Laptop')
# We force the 'Product' column to 'Title Case' (Capitalize first letter only).
df['Product'] = df['Product'].str.title()

# We also standardise 'Region' just in case
df['Region'] = df['Region'].str.title()

# 4. Handle Missing Data (Imputation)
# Issue A: Missing Payment Methods. We assume if it's blank, it was 'Cash'.
df['PaymentMethod'] = df['PaymentMethod'].fillna('Cash')

# Issue B: Missing Sales Amount.
# In finance, we cannot "guess" revenue. If sales data is missing, the record is invalid.
# We drop rows where 'SalesAmount' is empty (NaN).
df = df.dropna(subset=['SalesAmount'])

# 5. Verification
print(f"Final Clean Row Count: {len(df)}")
print("\nMissing Values Audit:")
print(df.isnull().sum()) # Should be all zeros now

# 6. Save the Clean File
# We save this as a NEW file. Never overwrite your raw source data.
df.to_csv('globaltech_clean_sales.csv', index=False)
print("\nSUCCESS: 'globaltech_clean_sales.csv' saved successfully.")
```
