```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# 1. Setup - Create 1000 rows
num_rows = 1000
data = []

products = ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Headset', 'Webcam']
regions = ['North', 'South', 'East', 'West', 'North', 'South'] # Duplicates to skew data
payment_methods = ['Credit Card', 'PayPal', 'Bank Transfer', 'Cash', None] # None creates missing data

# 2. Generate Data Loop
for i in range(num_rows):
    # Create Date (last 365 days)
    date = datetime.now() - timedelta(days=random.randint(0, 365))
    date_str = date.strftime('%Y-%m-%d')
    
    # Create Product (with occasional typos)
    prod = random.choice(products)
    if random.random() < 0.05: # 5% chance of typo
        prod = prod.upper() if random.random() < 0.5 else prod.lower()
        
    # Create Sales Amount (Price * Qty)
    qty = random.randint(1, 10)
    price = {'Laptop': 800, 'Mouse': 20, 'Monitor': 200, 'Keyboard': 50, 'Headset': 80, 'Webcam': 40}
    # Fix casing for price lookup
    base_price = price.get(prod.capitalize(), 50) 
    sales = base_price * qty
    
    # Introduce Nulls/Errors
    if random.random() < 0.03: # 3% chance sales is missing
        sales = np.nan
    
    region = random.choice(regions)
    if random.random() < 0.02: # 2% chance region is "N/A"
        region = "N/A"

    payment = random.choice(payment_methods)

    data.append([i + 1000, date_str, prod, region, qty, sales, payment])

# 3. Create DataFrame
df = pd.DataFrame(data, columns=['TransactionID', 'Date', 'Product', 'Region', 'Quantity', 'SalesAmount', 'PaymentMethod'])

# 4. Add intentional duplicates
df = pd.concat([df, df.iloc[:20]]) # Duplicate first 20 rows

# 5. Save to CSV
df.to_csv('globaltech_raw_sales.csv', index=False)
print("SUCCESS: 'globaltech_raw_sales.csv' has been created with 1,020 rows (including duplicates).")
```
